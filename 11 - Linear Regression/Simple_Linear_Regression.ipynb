```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('placement (1).csv')
```

```python
df.head()
```

```python
plt.scatter(df['cgpa'], df['package'])
plt.xlabel('CGPA')
plt.ylabel('Package (in Lacs)')
plt.title('CGPA vs Package')
```

```python
X = df.iloc[:,0:1]
y = df.iloc[:,-1]
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2, random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train, y_train)
```

```python
X_test
```

```python
lr.predict(X_test.iloc[2].values.reshape(1,1)) # LR expects data should be one row and one column so we reshape it
```

```python
plt.scatter(df['cgpa'], df['package'])
plt.plot(X_train, lr.predict(X_train), color='red') # Linear Regression Line
# The red line shown is best fit line
plt.xlabel('CGPA')
plt.ylabel('Package (in Lacs)')
plt.title('CGPA vs Package')
```

```python
m = lr.coef_ # Slope of the line
```

```python
c = lr.intercept_ # Y intercept
```

```python
# Line Equation
# y = mx + c

m * 8.58 + c
```

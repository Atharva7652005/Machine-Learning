```python
from IPython.display import HTML

html_code = """
<h1> 1. Numerical Data </h1>
<h2> a. Histogram </h2>
"""

HTML(html_code)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.hist(df['Age'])
# #plt.hist(df['Age'], bins=50)

# The bins parameter in histograms defines how data is grouped into intervals,
# or "bins", to create the visual representation of the data distribution

# Each bin represents a range of values, and the height of the
# bar corresponds to the number of data points falling within that range
```

```python
from IPython.display import HTML

html_code = """
<h2> b. Distplot </h2>
"""

HTML(html_code)
```

```python
# sns.distplot(df['Age']) # Deprecated
sns.displot(df['Age'], kde=True)
# ecdf : empirical cumulative distribution function
# kde : kernel density estimation is a line on displot which shows the the distribution of data

# displot() method is a function that provides access to several approached for visualizing univariate and bivariate distribution of data.
# This function like other functions in the Seaborn library allows the plotting
# of subsets of data defined by semantic mapping across multiple subplots.

#displots also helps to calculate skewness
```
